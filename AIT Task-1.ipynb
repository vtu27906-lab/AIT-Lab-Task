```python
# BFS and DFS for a graph with 3 nodes

graph = {
    0: [1, 2],
    1: [0],
    2: [0]
}

# BFS
def bfs(start):
    visited = []
    queue = [start]

    while queue:
        node = queue.pop(0)

        if node not in visited:
            visited.append(node)
            queue.extend(graph[node])

    return visited


# DFS
def dfs(node, visited=None):
    if visited is None:
        visited = []

    visited.append(node)

    for neighbour in graph[node]:
        if neighbour not in visited:
            dfs(neighbour, visited)

    return visited


print("BFS Traversal:", bfs(0))
print("DFS Traversal:", dfs(0))
```
